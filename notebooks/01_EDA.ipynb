```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import drive

drive.mount('/content/drive')

print("Libraries imported successfully!")
```

```python
import os

# Updated to match the exact folder name 'Fraud-ATO-Detection.' found in your Google Drive listing
DATA_PATH = '/content/drive/MyDrive/Fraud-ATO-Detection./data/creditcard.csv'

if not os.path.exists(DATA_PATH):
    print(f"Error: The path '{DATA_PATH}' does not exist.\n")
    print("Listing folders/files under '/content/drive/MyDrive/Fraud-ATO-Detection.' to locate your CSV:")
    try:
        for item in os.listdir('/content/drive/MyDrive/Fraud-ATO-Detection.'):
            print(f" - {item}")
    except Exception as e:
        print("Could not list directory contents:", e)
else:
    df = pd.read_csv(DATA_PATH)
    print("Dataset loaded successfully!")
```

```python
print("Dataset Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
print("Total duplicates:", df.duplicated().sum())
```

```python
class_counts = df['Class'].value_counts()

print("Legitimate transactions:", class_counts[0])
print("Fraudulent transactions:", class_counts[1])

print("\nPercentage distribution:")
print(df['Class'].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Class')

plt.title('Fraud vs Legitimate Transactions')
plt.xlabel('Transaction Class')
plt.ylabel('Number of Transactions')

plt.xticks([0, 1], ['Legitimate', 'Fraud'])

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(data=df, x='Class', y='Amount')

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Class')
plt.ylabel('Transaction Amount')

plt.xticks([0, 1], ['Legitimate', 'Fraud'])

plt.show()
```

```python
df.groupby('Class')['Amount'].agg(
    ['mean', 'median', 'min', 'max']
)
```

```python
plt.figure(figsize=(16, 12))

correlation = df.corr()

sns.heatmap(
    correlation,
    cmap='coolwarm',
    center=0,
    linewidths=0.1
)

plt.title('Feature Correlation Heatmap')

plt.show()
```

```python
import os

OUTPUT_PATH = '/content/drive/MyDrive/Fraud-ATO-Detection/figures'

os.makedirs(OUTPUT_PATH, exist_ok=True)

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Class')

plt.title('Fraud Distribution')

plt.savefig(
    f'{OUTPUT_PATH}/fraud_distribution.png',
    bbox_inches='tight'
)

plt.show()
```

```python
# Investigate duplicate records

duplicates = df[df.duplicated(keep=False)]

print("Total duplicate rows:", df.duplicated().sum())

print("\nDuplicate class distribution:")
print(duplicates['Class'].value_counts())

print("\nDuplicate percentage:")
print((df.duplicated().sum() / len(df)) * 100, "%")
```
